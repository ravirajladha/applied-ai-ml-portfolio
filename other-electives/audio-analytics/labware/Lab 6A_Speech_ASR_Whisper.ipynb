````python
import whisper
import librosa
import jiwer
import time

AUDIO_PATH = "audio.wav"   # replace with your own sample
REFERENCE_TEXT = "add the sum to the product of these three"

print("Loading Whisper (base, CPU)...")
model = whisper.load_model("base", device="cpu")
print("Model loaded.")
````

````python
# Load and inspect audio
audio, sr = librosa.load(AUDIO_PATH, sr=16000)
duration = len(audio) / sr
print(f"Audio duration: {duration:.2f}s | Sample rate: {sr} Hz")
````

````python
# Transcribe
start = time.time()
result = model.transcribe(AUDIO_PATH, language="en", fp16=False)
elapsed = time.time() - start

hypothesis = result["text"].strip()
print(f"Transcription time: {elapsed:.2f}s")
print(f"Hypothesis : {hypothesis}")
print(f"Reference  : {REFERENCE_TEXT}")
````

````python
# WER calculation (ties back to WER slide in M7)
wer = jiwer.wer(REFERENCE_TEXT.lower(), hypothesis.lower())
print(f"WER: {wer*100:.2f}%")

measures = jiwer.process_words(REFERENCE_TEXT.lower(), hypothesis.lower())
print(f"Substitutions: {measures.substitutions}, Deletions: {measures.deletions}, Insertions: {measures.insertions}")
````

## Summary

In this lab, you learned:

2. **WER**: Compute Word Error Rate against ground-truth reference
3. **Performance**: Clean speech typically yields low WER even with `base` checkpoint

### Key Takeaways
- Whisper generalizes well across accents/noise due to multitask, multilingual training
- WER degrades with noisy/accented speech — try LibriSpeech `dev-other` for comparison
- Larger checkpoints (small, medium) trade speed for accuracy
