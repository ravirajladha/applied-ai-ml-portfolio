POS Tagging using NLTK (Natural Language Toolkit)

```python
import nltk

# Download necessary models and resources
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

# Sample text
text = "Natural Language Processing enables machines to understand human language."

# Tokenize the sentence
tokens = nltk.word_tokenize(text)

# Apply POS tagging
tags = nltk.pos_tag(tokens)

# Display token and its POS tag
print("POS Tagging using NLTK:")
print(tags)
```

Named Entity Recognition with spaCy

```python
# ------------------------------------------
# Named Entity Recognition (NER) using spaCy
# ------------------------------------------

import spacy

# Load spaCy's small English model which includes NER
nlp = spacy.load("en_core_web_sm")

# Sample input text for NER
text = "Elon Musk founded SpaceX and Tesla. He lives in the United States."

# Process the text to extract entities
doc = nlp(text)

# ------------------------------------------
# Extract and Display Named Entities
# ------------------------------------------
print("Named Entities Found:")
for ent in doc.ents:
    print(f"{ent.text} ({ent.label_})")
```

Sentiment Analysis using NLTK

VADER Sentiment Analyzer from nltk.sentiment (Good for social media & general English)

```python
# Import the Natural Language Toolkit (nltk)
import nltk

# Download the VADER lexicon – a pre-built rule-based sentiment lexicon
# specifically designed for analyzing social media and general text sentiment
nltk.download('vader_lexicon')

# Import the VADER sentiment analyzer class
from nltk.sentiment import SentimentIntensityAnalyzer

# Initialize the VADER sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Define an example sentence to analyze
text = "Coursera's NLP course is EXCELLENT !!!"

# Get sentiment scores as a dictionary with 4 keys:
# - 'neg': negative sentiment score
# - 'neu': neutral sentiment score
# - 'pos': positive sentiment score
# - 'compound': overall sentiment score (ranges from -1 to +1)
scores = sia.polarity_scores(text)

# Print the full sentiment score dictionary
print("Sentiment Scores:", scores)
```

Text Generation using Hugging Face Transformers

```python
from transformers import pipeline

# Use a pre-trained language generation model
generator = pipeline("text-generation", model="gpt2")

# Generate text
prompt = "The future of natural language processing is"
output = generator(prompt, max_length=30, num_return_sequences=1)

print("Generated Text:")
print(output[0]['generated_text'])
```
